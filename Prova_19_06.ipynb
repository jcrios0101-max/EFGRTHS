```python
#2 Solicite ao usuário um número inteiro positivo.
#Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

n = int(input("Digite um número: "))

if n <= 0:
  print("digite um número")
else:
  soma = 0
  for i in range(1, n + 1):
    soma += i
  print(f"A soma de todos os números de 1 até {n} é: {soma}")
```

```python
#3 Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

numero = int (input("digite um numero: "))

if 1 <= numero <= 10:
  print(f"tabuada do {numero}:")
  for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
#4 Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

numero = int(input("Digite um número positivo: "))

if numero < 0:
  print("Eu disse positivo!")
elif numero == 0:
  print("O fatorial de 0 é 1.")
else:
  fatorial = 1
  for i in range(1, numero + 1):
    fatorial *= i
  print(f"O fatorial de {numero} é: {fatorial}")
```

```python
#5 Um caixa eletrônico possui notas de:
#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1
#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

a = int(input("digite o valor do saque: "))
cem = 0
cinquenta = 0
vinte = 0
dez = 0
cinco = 0
dois = 0
um = 0

while a > 0:
  if a >= 100:
     a -= 100
     cem += 1
  elif a >= 50:
    a -= 50
    cinquenta += 1
  elif a >= 20:
    a -= 20
    vinte += 1
  elif a >= 10:
    a -= 10
    dez += 1
  elif a >= 5:
    a -= 5
    cinco += 1
  elif a >= 2:
    a -= 2
    dois += 1
  elif a >= 1:
    a -= 1
    um += 1

if cem > 0:
  print(f"{cem}  nota(s) de R$100")
if cinquenta > 0:
  print(f"{cinquenta}  nota(s) de R$50")
if vinte > 0:
  print(f"{vinte}  nota(s) de R$20")
if cem > 0:
  print(f"{dez}  nota(s) de R$10")
if cem > 0:
  print(f"{cinco}  nota(s) de R$5")
if cem > 0:
  print(f"{dois}  nota(s) de R$2")
if cem > 0:
  print(f"{um}  nota(s) de R$1")
```
