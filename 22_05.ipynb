```python
##Escrever um programa que verifica as notas dos 4 bimestres e calcula a media do aluno

#nao permitir que o usuario digite notas menores que 0 e maiores que 10
#media menor que 5 - reprovado
#media maior ou igual a 5 e menor que 7 - recuperação
#media maior ou igual 7 - aprovado

#entrada

nota1 = float (input("digite a nota do bimestre 1: "))
nota2 = float (input("digite a nota do bimestre 2: "))
nota3 = float (input("digite a nota do bimestre 3: "))
nota4 = float (input("digite a nota do bimestre 4: "))

#Procesamento/saída

media = ((nota1+nota2+nota3+nota4)/4)

if media < 5 :

  print (f"media: {media} aluno reprovado")

elif media >= 5 and media < 7 :

  print (f"media: {media} aluno de recuperação")

else:

  print (f"media: {media} aluno aprovado")
```
